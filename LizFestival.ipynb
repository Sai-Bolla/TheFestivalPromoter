```python
import requests
from pprint import pprint
import time
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
#  Set up env key and api - TESTING
# https://app.ticketmaster.com/{package}/{version}/{resource}.json?apikey=**{API key}
TICKET_API_KEY = os.getenv("TICKET_API_KEY")
# params = {'apikey': TICKET_API_KEY}
BASE_URL = "https://app.ticketmaster.com/discovery/v2/events.json?"
EVENTS_ENDPOINT = F"apikey={TICKET_API_KEY}"
COUNTRY_ENDPOINT = f"countryCode=GB&apikey={TICKET_API_KEY}"

MUSIC_ENDPOINT = F"classificationName=music&dmaId=601&apikey={TICKET_API_KEY}"

full_path = f"{BASE_URL}{MUSIC_ENDPOINT}"
response = requests.get(full_path)

print(f"Status code: {response.status_code}")
print(f"URL Used: {response.url}")

data = response.json()
```
